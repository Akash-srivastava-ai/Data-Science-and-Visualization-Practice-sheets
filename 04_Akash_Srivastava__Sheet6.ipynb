```python
quantity = np.array([
    [12, 18, 25, 10],
    [20, 15, 30, 12],
    [18, 22, 16, 20],
    [25, 19, 24, 15],
    [14, 28, 20, 18],
    [22, 17, 26, 21]
])
price = np.array([150, 240, 90, 320])
revenue = quantity*price
daily_revenue = np.sum(revenue,axis=1)
product_revenue = np.sum(revenue,axis=0)
total_quantity = np.sum(quantity,axis=0)
print(daily_revenue)
print(product_revenue)
print(np.argmax(daily_revenue)+1)
print(np.argmax(total_quantity)+1)
new_price = np.where(total_quantity>100,price*0.9,price)
print(new_price)
print(np.sum(quantity*new_price))
```

Q38) replacing values with boundaries reduces extreme values and usually reduces variance and standard deviation

```python
np.random.seed(42)
a = np.random.normal(50,12,100)
print(np.sum((a<30)|(a>70)))
b = np.where(a<30,30,np.where(a>70,70,a))
print(np.mean(a))
print(np.mean(b))
print(np.var(a))
print(np.var(b))
print(np.std(a))
print(np.std(b))
print(np.min(a),np.max(a))
print(np.min(b),np.max(b))
```

Q39) ordinary addition gives NaN if a quarter is missing from any branch
fill_value=0 assumes that a missing record means zero earnings

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
a = branch_a + branch_b + branch_c
print(a)
print(set(branch_a.index)&set(branch_b.index)&set(branch_c.index))
print((branch_a['Q4']+branch_b['Q4']+branch_c['Q4']))
b = branch_a.add(branch_b,fill_value=0).add(branch_c,fill_value=0)
print(b)
```

Q40) missing records should not automatically become zero because no score may have been recorded

```python
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
print(s)
print(s[s.isnull()])
a = s[s.notnull() & (s>=70)]
print(a)
b = a.to_dict()
print(b)
c = pd.Series(b)
print(c)
print(c.equals(a))
```
